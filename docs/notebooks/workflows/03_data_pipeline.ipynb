# Unified data pipeline

Normalize records, transform fields, define a task, and compile a sample.

```python
from zynnova.data import (
    FieldSpec, MaterialSample, MaterialType, SequenceSource,
    TaskSpec, compile_sample, pipeline,
)

sample = MaterialSample(
    id='molecule-1', material_type=MaterialType.MOLECULAR,
    labels={'energy': -1.23}, features={'descriptor': 2.0},
)
task = TaskSpec.property_prediction(
    'energy-demo', targets=[FieldSpec('energy', 'labels.energy')],
    inputs=[FieldSpec('descriptor', 'features.descriptor')],
    representation='none',
)
task = task.__class__(
    name=task.name, kind=task.kind, inputs=task.inputs, targets=task.targets,
    conditions=task.conditions, structure=None,
    material_types=task.material_types,
)
compiled = compile_sample(sample, task, tensorize=False)
print(compiled.to_dict())
print(pipeline(SequenceSource([sample])).materialize())
```
