# Polymer semantic record and training views

Construct a minimal repeat-unit record using only `zynnova.structure.polymer`.

```python
from zynnova.structure.polymer import (
    ArchitectureNode, ArchitectureType, Atom, MolecularGraph,
    PolymerArchitecture, PolymerRecord, PolymerUnit, UnitRole,
    RepresentationSchema, make_view,
)

unit_graph = MolecularGraph(atoms=[Atom(6), Atom(6)], bonds=[])
unit = PolymerUnit(id='A', role=UnitRole.REPEAT, graph=unit_graph)
architecture = PolymerArchitecture(
    architecture_type=ArchitectureType.LINEAR,
    nodes=[ArchitectureNode(id='n0', unit_id='A')],
    sequence=['A'],
)
record = PolymerRecord(id='poly-001', units={'A': unit}, architecture=architecture)
record.validate()
schema = RepresentationSchema.fit([record])
print(record.id, schema.unit_vocabulary)
```

```python
for view_name in ('chemical', 'single_chain', 'multiscale', 'transformer'):
    try:
        view = make_view(record, view_name, schema=schema)
    except TypeError:
        view = make_view(record, view_name)
    print(view_name, type(view).__name__)
```
