# Structure and graph round-trip

Build a molecule, create a graph, validate geometry, and reconstruct the structure.

```python
import numpy as np
from zynnova import StructureData
from zynnova.structure.molecular import graph2stru, stru2graph

water = StructureData(
    atomic_numbers=np.array([8, 1, 1]),
    positions=np.array([[0., 0., 0.], [0.958, 0., 0.], [-0.240, 0.927, 0.]]),
)
graph = stru2graph(water, backend='python', neighbor_mode='radius')
graph.validate_geometry()
restored = graph2stru(graph)
print(graph.num_nodes, graph.num_edges)
print(np.max(np.abs(restored.positions - water.positions)))
```

```python
from zynnova.structure.crystal import stru2graph as crystal_to_graph

crystal = StructureData(
    atomic_numbers=np.array([14]),
    positions=np.array([[0., 0., 0.]]),
    cell=np.eye(3) * 5.43,
    pbc=np.array([True, True, True]),
)
periodic_graph = crystal_to_graph(crystal, backend='python', cutoff=3.0)
print(periodic_graph.graph_attrs)
```
