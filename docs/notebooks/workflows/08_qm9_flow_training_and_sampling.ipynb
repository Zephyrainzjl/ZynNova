# QM9 equivariant flow matching

This is a complete public training entry point. It uses the built-in configuration, external workspace, checkpoint, and early-stopping contracts.

Install `zynnova[ml-all]`. Dataset preparation and training can be computationally expensive, so execution is opt-in.

```python
try:
    from zynnova.ml import QM9FlowConfig, QM9FlowDataConfig, QM9FlowTrainConfig, train_qm9_flow
except ImportError as exc:
    print('Install the ML extras first:', exc)
    ML_READY = False
else:
    ML_READY = True
```

```python
RUN_TRAINING = False

if ML_READY and RUN_TRAINING:
    config = QM9FlowConfig(
        data=QM9FlowDataConfig(limit=5000),
        train=QM9FlowTrainConfig(epochs=25, workspace_root='zynnova-work'),
    )
    result = train_qm9_flow(config)
    print(result.best_checkpoint, result.best_metric)
```

## Production checklist

1. Set a dedicated `workspace_root` outside the installed package.
2. Start with a small `limit`, verify units and masks, then scale up.
3. Keep the generated `config.json`, `history.jsonl`, `best.pt`, and `last.pt`.
4. Evaluate on a held-out split before downstream screening or simulation.
