# Molecular dynamics and restart

A small ASE/EMT run illustrates the public configuration and output contract.

```python
from pathlib import Path
from zynnova.dynamics import (
    MDConfig, OutputConfig, RunConfig, SafetyConfig, VelocityConfig,
    create_classical_calculator, run_md,
)

RUN_MD = False
if RUN_MD:
    from ase.build import molecule
    atoms = molecule('H2')
    calculator = create_classical_calculator('emt')
    config = RunConfig(
        md=MDConfig(steps=20, timestep_fs=0.2, ensemble='nve'),
        velocities=VelocityConfig(temperature_K=100.0),
        output=OutputConfig(directory=Path('md-demo'), overwrite=True, checkpoint_interval=10),
        safety=SafetyConfig(check_interval=1),
    )
    result = run_md(atoms, calculator, config)
    print(result.status, result.completed_steps)
```
