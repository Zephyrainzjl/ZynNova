# Public visualization workflow

Visual rendering requires `zynnova[visualization]`. No internal renderer is used.

```python
import numpy as np
from zynnova import StructureData
from zynnova.visualization import available_backends, visualize_molecule

molecule = StructureData(
    atomic_numbers=np.array([1, 1]),
    positions=np.array([[0., 0., 0.], [0.74, 0., 0.]]),
)
print('Available:', available_backends())
RUN_VIEWER = False
if RUN_VIEWER:
    viewer = visualize_molecule(molecule, backend='py3dmol')
    display(viewer)
```
