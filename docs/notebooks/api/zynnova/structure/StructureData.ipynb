## Minimal public usage

The example below stays on the public API surface.

````python
import numpy as np

if API is not None:
    structure = API(
        atomic_numbers=np.array([8, 1, 1]),
        positions=np.array([
            [0.000, 0.000, 0.000],
            [0.958, 0.000, 0.000],
            [-0.240, 0.927, 0.000],
        ]),
    )
    print(structure.num_atoms, structure.positions.shape)
````

## Expected behavior and common failures

- `ValueError('atomic_numbers must have shape [N]')`
- `ValueError('positions must have shape [N, 3]')`
- `ValueError('cell must have shape [3, 3]')`
- `ValueError('pbc must have shape [3]')`
- `ValueError('Cannot compute fractional positions from a singular cell')`
- `ValueError('bonds must have shape [M, 2]')`
- `ValueError('bonds contain invalid atom indices')`
- `ValueError('bond_orders must have shape [M] and require bonds')`
- `ImportError('ASE is required; install zynnova[io]')`
- `ValueError(f'{name} must have shape [N]')`

For dataset downloads, model training, molecular dynamics, and external engines, first test with a small input and an explicit output directory.
