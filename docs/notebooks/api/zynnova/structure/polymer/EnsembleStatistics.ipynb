## Expected behavior and common failures

- `ValueError('crosslink_density cannot be negative')`
- `ValueError('number_of_chains must be positive')`
- `ValueError(f'composition references unknown units: {sorted(unknown)}')`
- `ValueError('composition fractions cannot be negative')`
- `ValueError(f'composition must sum to 1, got {total}')`
- `ValueError('transition_matrix shape must match transition_unit_order')`
- `ValueError('transition matrix references unknown units')`
- `ValueError('transition probabilities cannot be negative')`
- `ValueError('each transition matrix row must sum to 1')`
