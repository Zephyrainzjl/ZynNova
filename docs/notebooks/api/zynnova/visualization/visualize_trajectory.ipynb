## Expected behavior and common failures

- `ValueError('interval must be positive')`
- `ValueError('reps must be non-negative')`
- `ValueError("loop must be 'forward', 'backward', or 'backAndForth'")`
- `ValueError('trajectory contains no frames')`
- `ValueError(f'unknown structure kind: {resolved_kind}')`
- `ValueError('crystal trajectory visualization requires periodic frames')`
- `ValueError(f'trajectory frame {index} has {frame.num_atoms} atoms; expected {atom_count}')`
- `ValueError(f'trajectory frame {index} has a different atom ordering or composition')`
