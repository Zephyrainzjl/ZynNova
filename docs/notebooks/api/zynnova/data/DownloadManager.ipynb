## Expected behavior and common failures

- `DownloadError('all download mirrors failed:\n' + '\n'.join(errors))`

- `DownloadError(f'checksum mismatch for {path.name}: expected {expected}, got {actual}')`
- `DownloadError(f'unsupported archive type: {archive_type}')`
