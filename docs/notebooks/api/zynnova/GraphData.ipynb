## Minimal public usage

The example below stays on the public API surface.

````python
import numpy as np
from zynnova import StructureData
from zynnova.structure.molecular import stru2graph

if API is not None:
    structure = StructureData(
        atomic_numbers=np.array([1, 1]),
        positions=np.array([[0.0, 0.0, 0.0], [0.74, 0.0, 0.0]]),
    )
    graph = stru2graph(structure, backend="python")
    print(type(graph).__name__, graph.num_nodes, graph.num_edges)
````

## Expected behavior and common failures

- `ValueError('edge_index must have shape [2, E]')`
- `ValueError('edge_shift must have shape [E, 3]')`
- `ValueError('edge_vec must have shape [E, 3]')`
- `ValueError('edge_dist must have shape [E] or [E, 1]')`
- `ValueError('node_features first dimension must equal num_nodes')`
- `ValueError('edge_features first dimension must equal num_edges')`
- `ValueError('cell and pbc must have shapes [3,3] and [3]')`
- `ValueError('edge_index contains invalid node indices')`
- `ValueError('edge_vec is inconsistent with positions, cell, and edge_shift')`
- `ValueError('edge_dist is inconsistent with edge_vec')`
- `ValueError('fractional_positions must have shape [N,3]')`
- `ImportError('PyTorch Geometric is required; install zynnova[graph]')`

For dataset downloads, model training, molecular dynamics, and external engines, first test with a small input and an explicit output directory.
